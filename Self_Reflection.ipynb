```python
# ================================
# Exercise 3 — Self-Reflection
# Tool Used: ChatGPT (LLM via prompt engineering)
# ================================

# Original summary (BEFORE)
original_summary = """
Our team worked on improving the onboarding process this quarter.
We made several updates and fixed some issues that were affecting new users.
There are still some areas that need improvement, and we plan to continue refining the system.
Overall, progress has been positive.
"""

# Self-critique prompt used
self_reflection_prompt = """
You are an expert writing reviewer performing structured self-reflection.

Evaluate the summary using:
- Accuracy
- Clarity
- Audience appropriateness (executive audience)
- Length limit (≤120 words)
- Formatting (single paragraph)
- Focus (emphasize outcomes over process)

Identify weaknesses and then rewrite the summary so that it:
- Is ≤120 words
- Is one paragraph
- Includes one measurable outcome
- Includes one clear next step
- Uses professional tone
"""

# Requirements used for critique
requirements = {
    "Length limit": "≤120 words",
    "Audience": "Executive / professional",
    "Formatting": "Single paragraph",
    "Must include": "One measurable outcome + one next step",
    "Focus": "Outcomes over process",
    "Clarity": "No vague wording"
}

# Improved summary (AFTER)
improved_summary = """
This quarter, we improved the onboarding process, reducing new-user setup time by 25% and decreasing early churn by 12%.
These changes strengthened user activation and improved first-week engagement.
To build on this progress, we will conduct a targeted usability review within the next 30 days to identify additional friction points and prioritize further optimization.
"""

# Display results
print("===== ORIGINAL SUMMARY (BEFORE) =====")
print(original_summary)

print("\n===== SELF-CRITIQUE PROMPT USED =====")
print(self_reflection_prompt)

print("\n===== REQUIREMENTS USED FOR CRITIQUE =====")
for key, value in requirements.items():
    print(f"{key}: {value}")

print("\n===== IMPROVED SUMMARY (AFTER) =====")
print(improved_summary)
```
