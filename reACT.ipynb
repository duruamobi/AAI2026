```python
# =========================
# ACT (CODE v1)
# =========================
# Tool used: Google Colab (Python standard library only; re allowed)

import re

def classify_message(msg: str) -> str:
    """
    Returns exactly one lowercase label from:
    {payment, shipping, product support, general inquiry}
    No extra text.
    """

    if msg is None:
        return "general inquiry"

    msg = msg.strip()
    if not msg:
        return "general inquiry"

    m = msg.lower()

    payment_kw = [
        "payment", "pay", "paid", "charge", "charged", "billing",
        "refund", "money back", "card", "declined", "invoice"
    ]

    shipping_kw = [
        "shipping", "shipment", "track", "tracking", "delivery",
        "delivered", "arrive", "eta", "where is", "in transit"
    ]

    product_support_kw = [
        "broken", "damaged", "defective", "not working",
        "error", "issue", "problem", "warranty", "replace"
    ]

    def has_any(keywords):
        return any(k in m for k in keywords)

    # Priority rule:
    # product support > shipping > payment > general inquiry
    if has_any(product_support_kw):
        return "product support"
    if has_any(shipping_kw):
        return "shipping"
    if has_any(payment_kw):
        return "payment"

    return "general inquiry"


# =========================
# RUN / OBSERVE (v1)
# =========================
tests = [
    "",
    None,
    "Hi there",
    "My card was charged twice.",
    "Where is my package? Tracking hasn't updated.",
    "The treadmill arrived damaged and won't start.",
    "I need a refund and my order hasn't arrived.",
    "Do you offer team discounts?"
]

for t in tests:
    print(classify_message(t))


# =========================
# FIX / ITERATE (CODE v2)
# =========================
# Observed weakness:
# "I need a refund and my order hasn't arrived." was classified as shipping,
# but refund intent may need explicit override handling.
# Improvement: if refund/payment terms AND shipping delay terms appear together,
# classify as payment (financial resolution prioritized).

def classify_message_v2(msg: str) -> str:

    if msg is None:
        return "general inquiry"

    msg = msg.strip()
    if not msg:
        return "general inquiry"

    m = msg.lower()

    payment_kw = [
        "payment", "pay", "paid", "charge", "charged", "billing",
        "refund", "money back", "card", "declined", "invoice"
    ]

    shipping_kw = [
        "shipping", "shipment", "track", "tracking", "delivery",
        "delivered", "arrive", "eta", "where is", "in transit",
        "hasn't arrived", "hasnt arrived"
    ]

    product_support_kw = [
        "broken", "damaged", "defective", "not working",
        "error", "issue", "problem", "warranty", "replace"
    ]

    def has_any(keywords):
        return any(k in m for k in keywords)

    # Override rule: refund/payment + shipping delay → payment
    if has_any(payment_kw) and has_any(shipping_kw):
        return "payment"

    if has_any(product_support_kw):
        return "product support"
    if has_any(shipping_kw):
        return "shipping"
    if has_any(payment_kw):
        return "payment"

    return "general inquiry"


# Re-run same tests to demonstrate improvement
for t in tests:
    print(classify_message_v2(t))
```
